# GBO daily error summary

Scheduled on Posit Connect **every morning** (e.g. 7:00 AM): emails yesterday's app errors, grouped by bug, to GBO_ALERT_EMAIL (Ryker). Nothing is sent on a day with no errors.

First run once: `python3 -m migrations.migrate_app_errors`

Deploy from the repo root:

```
rsconnect deploy notebook error_summary_job.ipynb $(git ls-files '*.py') -n <your-connect-server>
```

Then in Connect: **Vars** -> DATABASE_URL, SMTP_HOST, SMTP_PORT, SMTP_USER, SMTP_PASSWORD, SMTP_FROM, GBO_ALERT_EMAIL (optional; defaults to Ryker); **Schedule** -> daily.
