# GBO weekly progress report emails

Scheduled on Posit Connect for **Mondays, 7:00 AM Central**: emails last week's report to every pitcher who threw.

Deploy from the repo root (bundles the app's Python files with it):

```
rsconnect deploy notebook weekly_reports_job.ipynb $(git ls-files '*.py') -n <your-connect-server>
```

Then in Connect: **Vars** -> add DATABASE_URL (same as the app), SMTP_HOST, SMTP_PORT, SMTP_USER, SMTP_PASSWORD, SMTP_FROM, GBO_APP_URL; **Schedule** -> weekly, Monday 7:00 AM.
